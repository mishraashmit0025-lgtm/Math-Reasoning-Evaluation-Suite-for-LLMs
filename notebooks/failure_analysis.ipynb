# Failure analysis
Loads `results/results.jsonl` (run `python -m mathbench eval ...` first) and breaks accuracy and failure modes down by model, topic and difficulty.

```python
import os, pandas as pd
if os.path.basename(os.getcwd()) == 'notebooks': os.chdir('..')
df = pd.read_json('results/results.jsonl', lines=True)
df.groupby('model').correct.mean()
```

```python
df.pivot_table(index='topic', columns='model', values='correct', aggfunc='mean').round(3)
```

```python
df.pivot_table(index='difficulty', columns='model', values='correct', aggfunc='mean').round(3)
```

```python
df[~df.correct].groupby(['model', 'failure']).size().unstack(fill_value=0)
```

```python
# tasks every model failed on: candidates for the hard split
solved = df.groupby('task_id').correct.any()
solved[~solved].index.tolist()[:20]
```
